```python
import duckdb
from pathlib import Path

# The database the pipelines load into: db/local_data.duckdb.
# Jupyter normally starts in the notebook's own folder (db/); otherwise assume the repo root.
DB_PATH = Path("local_data.duckdb")
if not DB_PATH.exists():
    DB_PATH = Path("db") / "local_data.duckdb"

# query helper, closes conn after query so that file is not locked
def q(sql: str):
    """Runs one query read-only and closes the connection, so the file is never left locked."""
    with duckdb.connect(str(DB_PATH), read_only=True) as conn:
        return conn.sql(sql).df()
```

```python
# what data is loaded? one schema per pipeline (its dataset_name)
q("""
    select
        table_schema,
        table_name
    from information_schema.tables
    order by table_schema, table_name
""")
```

```python
# Query loaded data: <dataset_name>.<table>, e.g. the orders from lab 010
q("""
    select
        *
    from lab_010_csv_basics.orders
""").head(10)
```
